### Messages

Message are the fundamental unit of context for models in langchain. they represent the input and output of models, carrying both the content and metadata needed to represent the state of conversation when interacting with an LLM . Message are objects that contain:

1. Role identifies the message type(e.g system,user)
2. content represents the actual content of the message(like text, images,audio documents etc)
3. Metadata optional fields such as response infotmation , message IDs, token usage

Langchain provides a standard message type that works across providers, ensuring consistent behaviour regardless of the model being called.

```python
import os
from langchain.chat_models import init_chat_model

# Ensure your GROQ_API_KEY environment variable is set
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

# Initialize using an active Groq model ID
model = init_chat_model("openai/gpt-oss-120b", model_provider="groq")

# Run the query
response = model.invoke("Please tell me what's AI and DS")
print(response.content)
```

### Text Prompt

Text prompts are strings ideal for starightforward generation tasks where you don't need to retain conversation histroy.

Use text prompts when:
1. You have a single standalone request
2. You don't need conversation histroy
3. You want minimal code complexity

```python
model.invoke("What is langchain , langraph, and RAG Vector database")
```

### Message Prompts

Alternatively you can pass in a list of messages to the model by providing a list of message objects

Message types:
1. System Message : Tell the model how to behave and context for interaction
2. Human Message : Represents user input and interactions the model
3. AI Message : Response generated by the model including text content tool calls and metadata
4. Tool Message : Represents the outputs of tool calls

```python
from langchain.messages import  SystemMessage ,HumanMessage,AIMessage

messages =[
    SystemMessage("You are a poetry expert"),
    HumanMessage("Write a poem on Artifical Intelligence")
]

response=model.invoke(messages)
response.content
```

```python
system_msg = SystemMessage("You are a helpful coding assistant.")

messages = [
    system_msg,
    HumanMessage("How do I CREATE A REST API?")

]

response = model.invoke(messages)
print(response.content)
```

## Detailed Information to the LLM through System Message

```python
from langchain.messages import  SystemMessage, HumanMessage

system_msg = SystemMessage(""" 
You are a senior AI Engineer with expertise in AI application.
Always provide code examples and explain your reasoning.
Be concise but through in your explanation.
""")

messages = [
    system_msg,
    HumanMessage("How do I build a production level Application and SAAS  Product?")

]
response = model.invoke(messages)
print(response.content)
```

```python
from langchain_core.messages import HumanMessage, AIMessage, ToolMessage
from langchain_core.tools import tool

# 1. Define the tool schema
@tool
def get_weather(location: str) -> str:
    """Get the current weather for a location."""
    return "sunny"

# 2. Bind the tool to your model
tools = [get_weather]
model_with_tools = model.bind_tools(tools)

# 3. Construct your message history
ai_msg = AIMessage(
    content="", # Best practice: pass empty string "" instead of [] if no text content
    tool_calls=[{
        "name": "get_weather",
        "args": {"location": "San Francisco"},
        "id": "call_1234"
    }]
)

tool_message = ToolMessage(
    content="sunny",
    tool_call_id="call_1234"
)

messages = [
    HumanMessage(content="What's the weather in San Francisco?"),
    ai_msg,
    tool_message,
]

# 4. Invoke using model_with_tools
response = model_with_tools.invoke(messages)
print(response.content)
```
